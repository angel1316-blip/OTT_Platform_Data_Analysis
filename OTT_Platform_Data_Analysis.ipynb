```python
import pandas as pd

# Step 1: Load the dataset and take 5,500 rows (within your 5,000-6,000 limit)
df = pd.read_csv('netflix_titles.csv')
df = df.head(5500)

# Step 2: Check basic details and missing values
print("Total rows and columns:", df.shape)
print("\nMissing values in each column:")
print(df.isnull().sum())

# Step 3: Fill missing values simply
df = df.fillna("Unknown")

# Question 1: How many Movies and TV Shows are available?
print("\n--- Question 1: Movies vs TV Shows ---")
print(df['type'].value_counts())

# Question 2: Which genre is most popular? (Top 5)
print("\n--- Question 2: Most Popular Genres ---")
print(df['listed_in'].value_counts().head(5))

# Question 3: Which year had the most content releases? (Top 5)
print("\n--- Question 3: Top Release Years ---")
print(df['release_year'].value_counts().head(5))

# Question 4: Which country produces the most content? (Top 5)
print("\n--- Question 4: Top Content Producing Countries ---")
print(df['country'].value_counts().head(5))

# Question 5: What are the most common movie durations? (Top 5)
print("\n--- Question 5: Common Durations ---")
movies = df[df['type'] == 'Movie']
print(movies['duration'].value_counts().head(5))

# Question 6: How many titles are rated TV-MA, PG-13, etc.?
print("\n--- Question 6: Content Ratings ---")
print(df['rating'].value_counts())
```

Project Conclusion: Netflix OTT Data Analysis

1. Content Type: Out of 5,500 titles analyzed, Netflix has more Movies (3,531) than TV Shows (1,969).
2. Popular Genres: Stand-Up Comedy and International Dramas are the most common categories.
3. Peak Release Years: Content production peaked in 2019 (950 titles) and 2020 (948 titles).
4. Top Countries: The United States (1,572) produces the most content, followed by India (619).
5. Movie Duration: Most movies run between 94 and 102 minutes.
6. Target Audience: Netflix mainly targets adult and teen viewers, with TV-MA (2,287) and TV-14 (1,409) being the top ratings.
